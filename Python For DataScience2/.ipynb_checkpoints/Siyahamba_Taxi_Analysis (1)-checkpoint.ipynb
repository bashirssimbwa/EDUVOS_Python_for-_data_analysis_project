# Siyahamba Taxi Association

**ITPFA0-44 Project — Question 1**

Siyahamba is a fictional minibus taxi association operating from five Gauteng ranks: Bree, Noord, Bara, Mabopane and Germiston. This notebook uses six months of trip-level data (1 July 2025 to 31 December 2025) in `siyahamba_taxi_trips.csv` to describe passenger loads, revenue, distance and peak-hour operations, so that fleet allocation and driver scheduling can be based on the data rather than on rank size alone.

The file must sit in the same folder as this notebook. Run the cells from top to bottom. Charts are saved in the `charts/` folder.


## Question 1 — NumPy (20 marks)

The passenger, fare, revenue and distance columns are read with NumPy. Pandas is only used from Question 2 onwards.


### 1.1 Passengers array: total, mean, minimum and maximum (4 marks)

`Passengers` is converted to a one-dimensional NumPy array. The four figures below are per trip, except the total, which is the sum of passengers across every trip.


```python
import numpy as np

data = np.genfromtxt(
    "siyahamba_taxi_trips.csv",
    delimiter=",",
    names=True,
    dtype=None,
    encoding="utf-8",
)

passengers = np.asarray(data["Passengers"], dtype=float)

print("Number of trips:", passengers.size)
print("Total passengers:", int(passengers.sum()))
print("Mean passengers per trip:", round(float(passengers.mean()), 2))
print("Minimum passengers per trip:", int(passengers.min()))
print("Maximum passengers per trip:", int(passengers.max()))
```

A typical trip carries about **12.43 passengers**. Loads range from **2** to **35**, and the association moved **28,636** passengers in **2,304** trips. The wide gap between the minimum and the maximum is a vehicle-size effect, which Question 4 looks at more closely: a Quantum does not load like a midibus.


### 1.2 Slicing trip revenue (4 marks)

The first slice is the first 15 revenue values. The second slice takes the last 50 revenue values and then keeps every fifth one (positions 0, 5, 10, … inside that block).


```python
trip_revenue = np.asarray(data["TripRevenue_ZAR"], dtype=float)

first_15 = trip_revenue[:15]
last_50 = trip_revenue[-50:]
every_fifth_of_last_50 = last_50[::5]

print("First 15 TripRevenue_ZAR values:")
print(first_15)
print()
print("Last 50 TripRevenue_ZAR values:")
print(last_50)
print()
print("Every fifth value from those last 50:")
print(every_fifth_of_last_50)
print("Count of every-fifth values:", every_fifth_of_last_50.size)
```

Slicing does not copy the business rules; it only picks out positions. `[:15]` is trips 1 to 15 in file order. `[-50:]` is the last 50 trips (late December). `[::5]` then steps through that block, so 50 values become 10. Those late-December revenues still vary a lot (from under R200 to over R900), which is the same vehicle mix showing up at the end of the file, not a single fare.


### 1.3 Standard deviation and variance of distance (4 marks)


```python
distance = np.asarray(data["Distance_km"], dtype=float)

# NumPy's default divides by N (the population formula), not by N - 1.
std_population = np.std(distance)
var_population = np.var(distance)

# The sample versions are shown only so they can be matched to pandas .describe() in Question 2.
std_sample = np.std(distance, ddof=1)
var_sample = np.var(distance, ddof=1)

print("Trips:", distance.size)
print("Mean distance (km):", round(float(distance.mean()), 2))
print("Population standard deviation (NumPy default, ddof=0):", round(float(std_population), 4))
print("Population variance (NumPy default, ddof=0):", round(float(var_population), 4))
print("Sample standard deviation (ddof=1):", round(float(std_sample), 4))
print("Sample variance (ddof=1):", round(float(var_sample), 4))
```

Most trips are about **30.6 km**, with a population standard deviation of **9.87 km** and a population variance of **97.48**. Routes are not all similar: Mamelodi and Protea Glen are about 12 km, while Randburg is about 44 km, so the spread is the route list, not noisy GPS.

NumPy’s default `np.std` and `np.var` divide by *N*. Pandas `.describe()` divides by *N − 1*, so its standard deviation is slightly higher (**9.8754**). Both are reported so the two answers are not mistaken for a calculation error.


### 1.4 Element-wise check of trip revenue (4 marks)

Trip revenue should be passengers multiplied by the fare on that same trip. The first 20 rows are checked with element-wise multiplication, and the same check is then applied to the full column.


```python
fare = np.asarray(data["Fare_ZAR"], dtype=float)

passengers_20 = passengers[:20]
fare_20 = fare[:20]
manual_revenue_20 = passengers_20 * fare_20
recorded_revenue_20 = trip_revenue[:20]
difference_20 = manual_revenue_20 - recorded_revenue_20

print("Passengers (first 20):")
print(passengers_20)
print("Fare_ZAR (first 20):")
print(fare_20)
print("Manual revenue (passengers x fare):")
print(manual_revenue_20)
print("Recorded TripRevenue_ZAR:")
print(recorded_revenue_20)
print("Difference (manual - recorded):")
print(difference_20)
print("First 20 rows match:", bool(np.allclose(manual_revenue_20, recorded_revenue_20)))

manual_revenue_all = passengers * fare
print("All 2,304 rows match:", bool(np.allclose(manual_revenue_all, trip_revenue)))
print("Largest absolute difference:", float(np.max(np.abs(manual_revenue_all - trip_revenue))))
```

Element-wise multiplication lines the two arrays up by position: trip *i* uses `passengers[i] * fare[i]`, not a single fare applied to every row. The difference is **0** on the first 20 trips and on all **2,304** trips. `TripRevenue_ZAR` is exactly passengers times fare. It is not a discounted or rounded figure, so later totals can be trusted.


### 1.5 Reshape the first 24 passenger counts (4 marks)


```python
first_24 = passengers[:24]
reshaped = first_24.reshape(6, 4)

# Reshaping does not change the values or how many there are. It only changes
# the arrangement. These are still the first 24 passenger counts; reshape(6, 4)
# lays them into 6 rows and 4 columns, filled row by row (C order). That is
# useful when a flat list needs to be read as a small table, for example six
# blocks of four trips. The product of the new dimensions must equal the
# length of the array (6 x 4 = 24). Any other shape, such as (5, 4), would
# raise an error because 20 is not 24.

print("Shape before:", first_24.shape)
print(first_24)
print("Shape after:", reshaped.shape)
print(reshaped)
print("Same values in the same order:", bool(np.array_equal(reshaped.ravel(), first_24)))
print("Row sums (passengers in each block of 4 trips):", reshaped.sum(axis=1))
```

The reshaped array is 6 × 4 and `ravel()` returns the original 24 numbers in the same order. Nothing was aggregated or dropped. Row sums are just a more convenient view of the same data.


## Question 2 — Pandas (25 marks)

### 2.1 Load the CSV, last 10 rows, dtypes and describe (5 marks)


```python
import pandas as pd

df = pd.read_csv("siyahamba_taxi_trips.csv")

print("Rows and columns:", df.shape)
print()
print("Last 10 rows:")
display(df.tail(10))
print("Data types:")
display(df.dtypes)
print("Summary statistics (.describe()):")
display(df.describe())
```

The file has **2,304 rows and 13 columns**. `describe()` reports the sample standard deviation (divide by *N − 1*). For `Distance_km` that is **9.875**, which matches `np.std(..., ddof=1)` from Question 1.3 and not the population figure of 9.873. Dates and times are still text at this point; they are converted to real dates in Question 3, when weekly grouping needs them.


### 2.2 Missing values in each column (3 marks)


```python
missing_counts = df.isna().sum()
missing_table = pd.DataFrame({
    "Missing values": missing_counts,
    "Percent of trips": (missing_counts / len(df) * 100).round(2),
})
display(missing_table)
print("Columns with at least one missing value:")
print(missing_table[missing_table["Missing values"] > 0])
```

Only two columns are incomplete. **`Rainfall_mm` is missing on 125 trips (5.43%)** and **`PassengerRating` is missing on 174 trips (7.55%)**. Trip identity, rank, route, vehicle, passengers, fare, revenue, distance and the peak-hour flag are complete, so the revenue totals do not depend on how the gaps are filled.


### 2.3 Handle missing rainfall and ratings (5 marks)


```python
# Rainfall_mm is missing on 125 trips and PassengerRating on 174.
# Both gaps are filled with the median. Rows are not dropped.
#
# Rainfall is skewed. Most recorded trips have 0 mm, so the median is 0.
# Filling with the mean (about 3.5 mm) would mark a blank log as a rainy
# trip and would exaggerate the rain.
#
# Ratings cluster around 3.7. The median is not dragged by a score of 1.8
# or 5.0 the way the mean is, so it stands in for a typical trip.
#
# Dropping the incomplete rows would throw away good revenue, rank and
# passenger figures just because a rating or a rainfall reading was blank.
# That would understate what the association earned. The operational
# columns used in the rest of this notebook were never missing.

rainfall_median = df["Rainfall_mm"].median()
rating_median = df["PassengerRating"].median()
print("Median Rainfall_mm used for fill:", rainfall_median)
print("Median PassengerRating used for fill:", rating_median)

df_clean = df.copy()
df_clean["Rainfall_mm"] = df_clean["Rainfall_mm"].fillna(rainfall_median)
df_clean["PassengerRating"] = df_clean["PassengerRating"].fillna(rating_median)

print("Missing values after the fill:")
print(df_clean.isna().sum())
```

After the fill, every column has **0** missing values and the row count is still **2,304**. Rainfall’s median is **0.0 mm**. The rating median is **3.7**. Later charts that do not use those two columns would look the same on `df` or on `df_clean`; the cleaned frame is used from here so that a rating or rainfall summary is not silently based on fewer trips.


### 2.4 Bara rank, peak hour only (4 marks)

`PeakHour` is 1 for a peak trip and 0 otherwise. In this file the flag lines up exactly with departures from **05:00–07:59** and **16:00–18:59**.


```python
bara_peak = df_clean[(df_clean["Rank"] == "Bara") & (df_clean["PeakHour"] == 1)].copy()

print("Bara peak-hour trips:", len(bara_peak))
print("Ranks in the filtered frame:", bara_peak["Rank"].unique().tolist())
print("PeakHour values in the filtered frame:", bara_peak["PeakHour"].unique().tolist())

with pd.option_context("display.max_rows", 250, "display.max_columns", 20):
    display(bara_peak)
```

The filter keeps **197** trips. Every remaining row is Bara and every remaining `PeakHour` value is 1, so both conditions were applied together rather than one after the other by accident.


### 2.5 Revenue and average passengers by vehicle type (4 marks)


```python
by_vehicle = (
    df_clean.groupby("VehicleType")
    .agg(
        Trips=("TripID", "count"),
        Total_TripRevenue_ZAR=("TripRevenue_ZAR", "sum"),
        Average_Passengers=("Passengers", "mean"),
    )
    .sort_values("Total_TripRevenue_ZAR", ascending=False)
)
by_vehicle["Average_Passengers"] = by_vehicle["Average_Passengers"].round(2)
by_vehicle["Total_TripRevenue_ZAR"] = by_vehicle["Total_TripRevenue_ZAR"].round(2)
display(by_vehicle)
```

The Quantum fleet produces the most revenue in total (**R390,101.50**) only because it runs most of the trips (**1,503**). It actually carries the fewest passengers per trip (**9.71**). The midibus is the opposite: **224** trips, but **23.89** passengers and the highest revenue per departure. The Sprinter sits between them. This is the fleet-allocation result in one table: more Quantums do not automatically mean a better peak-hour plan.


### 2.6 Estimated fuel cost and the five longest trips (4 marks)

`FuelCost_Est` is 25% of `TripRevenue_ZAR`, as specified. It is an estimate for comparison, not a measured fuel bill.


```python
df_clean["FuelCost_Est"] = df_clean["TripRevenue_ZAR"] * 0.25

longest_trips = df_clean.nlargest(5, "Distance_km")[
    [
        "TripID",
        "Date",
        "DepartureTime",
        "Rank",
        "Route",
        "VehicleType",
        "Passengers",
        "Distance_km",
        "TripRevenue_ZAR",
        "FuelCost_Est",
    ]
]
display(longest_trips)
```

The five longest trips are all **Noord–Randburg**, between **46.0 km and 46.3 km**. Distance does not fix the revenue: the same corridor produces R276 on an 8-passenger Quantum and R759 on a 22-passenger midibus. A long route only pays well when the vehicle is full. The 25% fuel estimate moves with revenue, so an empty long trip costs the association twice: a high kilometre reading and a weak fare take.


## Question 3 — Weekly revenue line chart (20 marks)

### 3.1 to 3.4 Weekly total trip revenue


```python
import matplotlib.pyplot as plt
from pathlib import Path

# Dates are converted here, after Question 2.1 has already shown the original text dtypes.
df_clean["Date"] = pd.to_datetime(df_clean["Date"])

charts = Path("charts")
charts.mkdir(exist_ok=True)

# Week ending Sunday, so each point is one operating week.
weekly_revenue = (
    df_clean.set_index("Date")
    .resample("W-SUN")["TripRevenue_ZAR"]
    .sum()
)

# Navy line on a white background stays easy to read when the notebook is printed.
# Gold markers mark each week as its own point, so a single high or low week
# is not lost in the line. The figure is wide because there are 27 weeks.
# Light gridlines make neighbouring weeks easier to compare without hiding the series.
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(
    weekly_revenue.index,
    weekly_revenue.values,
    color="#1B3A4B",
    marker="o",
    markersize=6,
    markerfacecolor="#E3A008",
    markeredgecolor="#1B3A4B",
    linewidth=2,
)
ax.set_title("Siyahamba Taxi Association — total weekly trip revenue (July–December 2025)")
ax.set_xlabel("Week ending")
ax.set_ylabel("Total trip revenue (ZAR)")
ax.grid(True, axis="y", linestyle="--", alpha=0.4)
ax.set_ylim(0, weekly_revenue.max() * 1.15)
fig.autofmt_xdate()
fig.tight_layout()
fig.savefig(charts / "weekly_revenue.png", dpi=150, bbox_inches="tight")
plt.show()

print("Weeks plotted:", len(weekly_revenue))
print(weekly_revenue.round(2).to_string())
print("Saved: charts/weekly_revenue.png")
```

Weekly revenue for a full week mostly sits between about **R26,000 and R34,000**. The highest full week ends **7 December 2025 (R35,992)**. The last point, the week ending **4 January 2026**, drops to **R13,022.50** because the data stop on **31 December**. That week contains only three days (29–31 December), not a collapse in demand. The first point is also a short week: the file starts on Tuesday 1 July. The gold markers are there so those incomplete weeks can be seen as single points rather than as a trend by themselves.

The chart is saved as `charts/weekly_revenue.png`.


## Question 4 — Five charts (25 marks)

Each chart has a title, axis labels, and a short reading of what it means for operations. The five charts are different views: rank revenue, route revenue, peak load by vehicle, the monthly trend, and the spread of revenue by vehicle.


### Chart 1 — Total revenue by rank


```python
rank_revenue = (
    df_clean.groupby("Rank")["TripRevenue_ZAR"]
    .sum()
    .sort_values(ascending=False)
)
rank_trips = df_clean.groupby("Rank")["TripID"].count().reindex(rank_revenue.index)

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(rank_revenue.index, rank_revenue.values, color="#1F7A6A")
ax.set_title("Total trip revenue by rank, July–December 2025")
ax.set_xlabel("Rank")
ax.set_ylabel("Total trip revenue (ZAR)")
ax.grid(True, axis="y", linestyle="--", alpha=0.4)
ax.set_ylim(0, rank_revenue.max() * 1.18)
for bar, trips in zip(bars, rank_trips):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 1500,
        f"R{bar.get_height():,.0f}\n{int(trips)} trips",
        ha="center",
        va="bottom",
        fontsize=8,
        color="#1B3A4B",
    )
fig.tight_layout()
fig.savefig(charts / "q4_revenue_by_rank.png", dpi=150, bbox_inches="tight")
plt.show()
```

Bree earns the most (R168,644 from 463 trips) and Mabopane the least (R146,718 from 461 trips), even though the ranks run a similar number of trips. Bara actually has the most departures (477) and still sits near the bottom on revenue, so rank performance should be judged on route mix and fare rather than on how busy the rank looks.


### Chart 2 — Total revenue by route


```python
route_revenue = (
    df_clean.groupby("Route")["TripRevenue_ZAR"]
    .sum()
    .sort_values(ascending=True)
)

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(route_revenue.index, route_revenue.values, color="#1B3A4B")
ax.set_title("Total trip revenue by route, July–December 2025")
ax.set_xlabel("Total trip revenue (ZAR)")
ax.set_ylabel("Route")
ax.grid(True, axis="x", linestyle="--", alpha=0.4)
for i, value in enumerate(route_revenue.values):
    ax.text(value + 400, i, f"R{value:,.0f}", va="center", fontsize=8, color="#5C6B73")
ax.set_xlim(0, route_revenue.max() * 1.16)
fig.tight_layout()
fig.savefig(charts / "q4_revenue_by_route.png", dpi=150, bbox_inches="tight")
plt.show()
```

**Noord–Randburg** is the strongest corridor at about **R73,175**, ahead of Bree–Diepsloot and Bara–Orlando, while Noord–Mamelodi (about **R27,044**) and Bara–Protea Glen (about **R36,771**) earn the least. Those two are short, low-fare shuttles (R16.50 and R17), so they need frequent small-vehicle turnarounds rather than the same fleet as Randburg.


### Chart 3 — Average passengers by vehicle, peak versus off-peak


```python
vehicle_order = ["Toyota Quantum", "Mercedes Sprinter", "Midibus"]
peak_load = (
    df_clean.groupby(["VehicleType", "PeakHour"])["Passengers"]
    .mean()
    .unstack()
    .reindex(vehicle_order)
)

x = np.arange(len(vehicle_order))
width = 0.36

fig, ax = plt.subplots(figsize=(9, 5))
off_bars = ax.bar(
    x - width / 2,
    peak_load[0],
    width,
    label="Off-peak",
    color="#1B3A4B",
)
peak_bars = ax.bar(
    x + width / 2,
    peak_load[1],
    width,
    label="Peak hour (05:00–07:59 and 16:00–18:59)",
    color="#E3A008",
)
ax.set_title("Average passengers per trip by vehicle type and peak period")
ax.set_xlabel("Vehicle type")
ax.set_ylabel("Average passengers per trip")
ax.set_xticks(x)
ax.set_xticklabels(vehicle_order)
ax.legend(frameon=False)
ax.grid(True, axis="y", linestyle="--", alpha=0.4)
ax.set_ylim(0, 38)
for bars in (off_bars, peak_bars):
    for bar in bars:
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height() + 0.4,
            f"{bar.get_height():.1f}",
            ha="center",
            va="bottom",
            fontsize=9,
        )
fig.tight_layout()
fig.savefig(charts / "q4_peak_passengers_by_vehicle.png", dpi=150, bbox_inches="tight")
plt.show()

print("Average passengers:")
display(peak_load.round(2))
```

Every vehicle loads harder in the peak, but a peak midibus averages about **30.7** passengers, close to the highest load in the file (35), while a peak Quantum averages only about **12.1**. The scarce larger vehicles should be placed on peak departures first, because a midday midibus wastes the only fleet that can clear a full morning or afternoon queue.


### Chart 4 — Monthly trip revenue


```python
monthly_revenue = df_clean.groupby(df_clean["Date"].dt.to_period("M"))["TripRevenue_ZAR"].sum()
monthly_revenue.index = monthly_revenue.index.astype(str)
month_labels = ["Jul 2025", "Aug 2025", "Sep 2025", "Oct 2025", "Nov 2025", "Dec 2025"]

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(
    month_labels,
    monthly_revenue.values,
    color="#C4492C",
    marker="o",
    markersize=8,
    linewidth=2,
)
ax.set_title("Total monthly trip revenue, July–December 2025")
ax.set_xlabel("Month")
ax.set_ylabel("Total trip revenue (ZAR)")
ax.grid(True, axis="y", linestyle="--", alpha=0.4)
ax.set_ylim(0, monthly_revenue.max() * 1.2)
for label, value in zip(month_labels, monthly_revenue.values):
    ax.text(label, value + 2500, f"R{value:,.0f}", ha="center", fontsize=8, color="#1B3A4B")
fig.tight_layout()
fig.savefig(charts / "q4_monthly_revenue.png", dpi=150, bbox_inches="tight")
plt.show()
```

Monthly revenue stays between about **R119,000 and R130,000** from July to November, then rises to **R148,423** in December, when passengers per trip also rise to about **13.9**. November is the soft month (**R119,402**), so the December jump is the planning exception, and the incomplete final week on the weekly chart should not be used to cancel that extra capacity.


### Chart 5 — Spread of trip revenue by vehicle type


```python
box_data = [
    df_clean.loc[df_clean["VehicleType"] == vehicle, "TripRevenue_ZAR"].values
    for vehicle in vehicle_order
]
box_colours = ["#5C6B73", "#1F7A6A", "#E3A008"]

fig, ax = plt.subplots(figsize=(9, 5))
box = ax.boxplot(
    box_data,
    tick_labels=vehicle_order,
    patch_artist=True,
    showfliers=False,
    medianprops={"color": "#1B3A4B", "linewidth": 2},
)
for patch, colour in zip(box["boxes"], box_colours):
    patch.set_facecolor(colour)
    patch.set_alpha(0.85)
ax.set_title("Spread of trip revenue by vehicle type")
ax.set_xlabel("Vehicle type")
ax.set_ylabel("Trip revenue (ZAR)")
ax.grid(True, axis="y", linestyle="--", alpha=0.4)
fig.tight_layout()
fig.savefig(charts / "q4_revenue_boxplot_by_vehicle.png", dpi=150, bbox_inches="tight")
plt.show()

revenue_spread = df_clean.groupby("VehicleType")["TripRevenue_ZAR"].describe().reindex(vehicle_order)
display(revenue_spread.round(2))
```

The revenue boxes barely overlap: a typical Quantum trip earns much less than a Sprinter trip, and the midibus sits higher again, with a mean of about **R648**. Putting another Quantum on the road adds a trip in the low box, whereas rostering a midibus or a Sprinter onto a peak Randburg, Diepsloot or Ga-Rankuwa departure moves that departure into a higher revenue band.


## Question 5 — What management should take from the charts (10 marks)

Over July to December 2025, Siyahamba ran 2,304 trips, carried 28,636 passengers and collected R771,642.50 in fare revenue. The five ranks are almost equally busy, so a rank should not be judged on trip counts alone. Revenue follows the route. Bree earns the most, and Noord–Randburg is the strongest single corridor. Noord–Mamelodi and Bara–Protea Glen earn the least per trip because the fares are only R16.50 and R17. Those short routes suit a Quantum that can turn around quickly.

Peak hour is the clearer signal for scheduling. Departures from 05:00–07:59 and 16:00–18:59 are 38% of trips, but they carry about 16 passengers rather than about 10 and produce nearly half of all revenue. Ratings are a little lower in the peak, which fits crowding rather than one weak rank. The Toyota Quantum runs about two-thirds of trips and earns the least per departure: about 10 passengers off-peak and about 12 in the peak. A peak midibus averages about 31 passengers, and Sprinters also run close to full at those hours. Larger vehicles should go to Randburg, Diepsloot, Ga-Rankuwa and Soweto in the peak, with Quantums kept on Protea Glen and Mamelodi.

December breaks an otherwise steady half-year. Revenue rose to about R148,400, while November was the quietest month, so December needs extra capacity. The last point on the weekly chart is not a collapse in demand. That week only contains 29 to 31 December.


```python
# Supporting figures quoted in the Question 5 summary, calculated from the cleaned data.
peak = df_clean["PeakHour"] == 1
print("Trips:", len(df_clean))
print("Passengers:", int(df_clean["Passengers"].sum()))
print("Revenue (ZAR):", round(float(df_clean["TripRevenue_ZAR"].sum()), 2))
print("Peak share of trips:", round(float(peak.mean()), 4))
print("Peak share of revenue:", round(float(df_clean.loc[peak, "TripRevenue_ZAR"].sum() / df_clean["TripRevenue_ZAR"].sum()), 4))
print("Avg passengers peak / off-peak:", round(float(df_clean.loc[peak, "Passengers"].mean()), 2), "/", round(float(df_clean.loc[~peak, "Passengers"].mean()), 2))
print("Avg rating peak / off-peak:", round(float(df_clean.loc[peak, "PassengerRating"].mean()), 2), "/", round(float(df_clean.loc[~peak, "PassengerRating"].mean()), 2))
print("December revenue:", round(float(df_clean.loc[df_clean["Date"].dt.month == 12, "TripRevenue_ZAR"].sum()), 2))
print("December avg passengers:", round(float(df_clean.loc[df_clean["Date"].dt.month == 12, "Passengers"].mean()), 2))
print("November revenue:", round(float(df_clean.loc[df_clean["Date"].dt.month == 11, "TripRevenue_ZAR"].sum()), 2))
```
