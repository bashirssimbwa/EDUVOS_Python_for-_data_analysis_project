```python
"""
This application demonstrates how to capture user input, convert numerical
datatypes into string data, and concatenate text values together.
It fulfills the requirements for Question 2 and Question 3 of the practice quiz.
"""

# Prompt the user to type in a whole number and instantly convert it to an integer type
user_integer = int(input("Enter an integer value: "))

# Convert the numerical integer data back into a text-based string format
integer_as_string = str(user_integer)

# Prompt the user to type in their name and store it directly as a text string
user_name = input("Enter your name: ")

# Link the user's name string and the converted integer string back-to-back
combined_output = user_name + integer_as_string

# Print out the final combined textual result onto the user's display
print("The combined result is:", combined_output)
```
