### Filling in Missing Data
- Transforming Data lUsing a Function or Mapping

```python
data = pd.DataFrame({'food': ['bacon', 'pulled pork', 'Pastorami', 'corned beef', 'Bacon', 
                             'pastorami', 'honey ham', 'nova lox'], 'ounces':[4,3,12,6,7.5,8,5,6]})
data
```

```python
meat_to_animal = {
    'bacon': 'pig',
    'pulled pork': 'pig',
    'pastorami': 'cow',
    'corned beef': 'cow',
    'honey ham': 'pig',
    'nova lox': 'salmon'
}

meat_to_animal
```

```python
lowercase = data['food'].str.lower()
lowercase
```

```python
data['animal'] = lowercase.map(meat_to_animal)

data
```
