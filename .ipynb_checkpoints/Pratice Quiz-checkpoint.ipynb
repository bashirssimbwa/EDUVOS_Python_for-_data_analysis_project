```python
# ==============================================================================
# COMPREHENSIVE PYTHON APPLICATION: STUDENT COURSE & FEE MANAGEMENT SYSTEM
# Demonstrating: Lists, Tuples, Dictionaries, and Program Control Structures
# ==============================================================================

#Student Number: EDUV10064199, 
#Student NAme: SSIMBWA BASHIR

def main():
    # --------------------------------------------------------------------------
    # 1. DATA STRUCTURE INITIALIZATION
    # --------------------------------------------------------------------------
    
    # DICTIONARY: Stores the baseline course catalog maps (Course ID -> Details)
    # The value is a nested dictionary mapping properties like names and baseline fees.
    course_catalog = {
        "COS101": {"name": "Introduction to Programming", "fee": 1200.00},
        "DAT201": {"name": "Data Structures & Algorithms", "fee": 1500.00},
        "NET301": {"name": "Advanced Networking Systems", "fee": 1800.00},
        "SEC401": {"name": "Cybersecurity Fundamentals", "fee": 2000.00}
    }
    
    # LIST: Tracks student registration data records sequentially.
    # This list will store individual immutable tuples representing transaction logs.
    registration_log = []
    
    # --------------------------------------------------------------------------
    # 2. PROGRAM CONTROL STRUCTURE: WHILE LOOP FOR MAIN MENU INTERACTION
    # --------------------------------------------------------------------------
    print("=== Welcome to the Eduvos Student Course Management System ===")
    
    is_running = True
    while is_running:
        print("\n--- MAIN MENU ---")
        print("1. View Available Course Catalog")
        print("2. Register a Student for a Course")
        print("3. View Session Registration Summary Log")
        print("4. Calculate Fees and Checkout")
        print("5. Exit Application")
        
        choice = input("Please select an option (1-5): ").strip()
        
        # --------------------------------------------------------------------------
        # 3. CONTROL STRUCTURES: IF-ELIF-ELSE CONDITIONAL STATEMENTS
        # --------------------------------------------------------------------------
        if choice == "1":
            # FOR LOOP: Iterating and displaying contents from the Dictionary structure
            print("\n--- AVAILABLE COURSE CATALOG ---")
            print(f"{'Course ID':<10} | {'Course Name':<30} | {'Fee (ZAR)':<10}")
            print("-" * 58)
            
            for code, details in course_catalog.items():
                # Accessing deep inner items from nested dictionary containers
                print(f"{code:<10} | {details['name']:<30} | R{details['fee']:<.2f}")
                
        elif choice == "2":
            print("\n--- REGISTER FOR A COURSE ---")
            student_id = input("Enter Student ID (e.g., ST123): ").strip().upper()
            course_code = input("Enter Course Code to join (e.g., COS101): ").strip().upper()
            
            # CONTROL STRUCTURE: Checking membership keys within a dictionary
            if course_code in course_catalog:
                course_info = course_catalog[course_code]
                
                # TUPLE: Immutable record structured for data tracking persistence.
                # Tuples prevent transactional record values from accidental mutation later.
                registration_record = (student_id, course_code, course_info['name'], course_info['fee'])
                
                # LIST OPERATION: Appending the tuple structure directly into our list sequence
                registration_log.append(registration_record)
                print(f"Success: Student {student_id} registered for {course_code}!")
            else:
                print("Error: The requested course code does not exist in the catalog.")
                
        elif choice == "3":
            print("\n--- CURRENT REGISTRATION LOG ---")
            # CONTROL STRUCTURE: Simple boolean condition checking if the list tracking sequence is empty
            if not registration_log:
                print("No registrations have been logged during this session.")
            else:
                # FOR LOOP: Unpacking tuple indices contained sequentially inside the tracking list
                for index, record in enumerate(registration_log, start=1):
                    # Unpacking tuple items directly into meaningful descriptive variables
                    std_id, c_code, c_name, fee = record
                    print(f"[{index}] Student ID: {std_id} | Course: {c_code} - {c_name} (R{fee:.2f})")
                    
        elif choice == "4":
            print("\n--- FINANCIAL SUMMARY & ACCUMULATION ---")
            if not registration_log:
                print("No active courses registered to calculate financials.")
            else:
                total_cost = 0.0
                
                # FOR LOOP: Iterating through list components to accumulate total sum aggregates
                for record in registration_log:
                    # Isolating the precise index variable element from our tracked tuples
                    _, _, _, fee = record
                    total_cost += fee
                
                print(f"Total Base Fees: R{total_cost:.2f}")
                
                # CONTROL STRUCTURE: Evaluating compound numbers to award conditional discounts
                if total_cost > 3000:
                    discount = total_cost * 0.10  # 10% multi-course promotional discount
                    final_total = total_cost - discount
                    print(f"Multi-course tier discount achieved (10%): -R{discount:.2f}")
                    print(f"Final Adjusted Balance Due: R{final_total:.2f}")
                else:
                    print(f"Final Adjusted Balance Due: R{total_cost:.2f}")
                    
        elif choice == "5":
            # Toggling flag to False breaks loop iteration control structures cleanly
            print("\nThank you for using the Eduvos Course Management System. Goodbye!")
            is_running = False
            
        else:
            print("Invalid input choice. Please select a valid option from the menu hierarchy (1-5).")

# Guard safety pattern executing entry execution parameters cleanly
if __name__ == "__main__":
    main()
```
