```python
# ==============================================================================
# PYTHON OPERATORS SHOWCASE APPLICATION
# Demonstrating: Arithmetic, Comparison, and Logical Operators (3 examples each)
# ==============================================================================

# --- TYPE 1: ARITHMETIC OPERATORS ---
# Used to perform common mathematical calculations

# Example 1: Addition (+)
sum_result = 15 + 5
print("Arithmetic - Addition (15 + 5):", sum_result)

# Example 2: Modulus (%) - Returns the division remainder
remainder_result = 17 % 4
print("Arithmetic - Modulus / Remainder (17 % 4):", remainder_result)

# Example 3: Exponentiation (**) - Raises a number to a power
power_result = 2 ** 3
print("Arithmetic - Exponentiation (2 ** 3):", power_result)


# --- TYPE 2: COMPARISON OPERATORS ---
# Used to compare two values, evaluating to either True or False

# Example 1: Greater than (>)
is_greater = 10 > 3
print("Comparison - Greater Than (10 > 3):", is_greater)

# Example 2: Equal to (==)
is_equal = (20 == 25)
print("Comparison - Equal To (20 == 25):", is_equal)

# Example 3: Not equal to (!=)
is_not_equal = (5 != 8)
print("Comparison - Not Equal To (5 != 8):", is_not_equal)


# --- TYPE 3: LOGICAL OPERATORS ---
# Used to combine conditional statements

# Example 1: Logical 'and' - Returns True if both conditions are True
and_result = (5 > 2) and (10 < 20)
print("Logical - AND ((5 > 2) and (10 < 20)):", and_result)

# Example 2: Logical 'or' - Returns True if at least one condition is True
or_result = (5 > 10) or (10 < 20)
print("Logical - OR ((5 > 10) or (10 < 20)):", or_result)

# Example 3: Logical 'not' - Reverses the Boolean result
not_result = not(5 > 10)
print("Logical - NOT (not(5 > 10)):", not_result)
```
