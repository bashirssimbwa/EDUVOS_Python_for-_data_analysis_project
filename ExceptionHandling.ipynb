```python
# SSIMBWA BASHIR EDUV10064199
# Simple Banking Simulator, Including Exception Handling
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance
    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"Successfully deposited R{amount}")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print(f"Successfully withdrawn R{amount} ")
        else:
            print("Insufficient Funds")
#Execution Loop
if __name__ == "__main__":
    print("====Welcome to our Banking App====")
    #Account Holdername
    while True:
        name = input("Enter your name Please!")
        if not name:
            print("Enter your name!")
            continue
        else: 
            break     
    #User Defined Function to capture Bad Numerical Inputs from Users
    def get_amount(prompt):#Non Negative Inputs, Empty inputs and Float values for Money 
        while True:
            try:
                amount = float(input(prompt))
                if not amount:
                    print("Amount cant be left blank")
                    continue
                if amount <= 0:
                    print("Ammount must be greater than Zero")
                    continue
                return amount
            except ValueError:
                print("Invalid Input")           
    #Initial Deposit From the User
    initial_deposit = get_amount("Please Enter your First deposit:  ")
    user_account = BankAccount(account_holder=name, balance=initial_deposit)
    while True:
        print("\n What do u want to do?")
        print("\n Select Options below \n 1. Make Deposit \n 2. Make Withdraw  \n 3.  Balance Inquiry \n 4. Exit Service")
        user_input = input("\n Select Options 1-4: ")
        if user_input == "1":
            amount = get_amount("Enter the amount to deposit in ZAR:")
            user_account.deposit(amount)
        elif user_input == "2":
            amount = get_amount("Enter the amount to withdraw in ZAR: ")
            user_account.withdraw(amount)    
        elif user_input == "3":
            print(f"{user_account.account_holder},  R{user_account.balance}")
        elif user_input == "4":
            print("===Thank U for Banking with US===== ")
            break
        else:
            print("Wrong Selection")
```
